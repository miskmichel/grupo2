## Contexto

Uma empresa de telecomunicações está perdendo clientes todo mês (esse fenômeno se chama **churn**) e quer agir de forma preventiva: identificar, com antecedência, quais clientes têm maior probabilidade de cancelar o serviço, para que o time de retenção possa agir antes que isso aconteça.

Você recebeu uma base histórica de clientes, cada um com um conjunto de características (dados demográficos, serviços contratados, forma de pagamento, tempo de contrato) e a informação se aquele cliente cancelou ou não (`Churn`).

## Objetivo

Construir o **melhor modelo de classificação binária possível** para prever `Churn` (Yes/No), dentro do tempo disponível.

Não existe "a resposta certa" — existe o melhor resultado que seu trio conseguir defender com uma métrica de avaliação apropriada.

## O dataset

Arquivo: `data/Telco-Customer-Churn.csv` — 7.043 clientes, 21 colunas.

| Coluna | Descrição |
|---|---|
| `customerID` | identificador único (não é feature) |
| `gender`, `SeniorCitizen`, `Partner`, `Dependents` | dados demográficos |
| `tenure` | meses de contrato |
| `PhoneService`, `MultipleLines`, `InternetService`, `OnlineSecurity`, `OnlineBackup`, `DeviceProtection`, `TechSupport`, `StreamingTV`, `StreamingMovies` | serviços contratados |
| `Contract`, `PaperlessBilling`, `PaymentMethod` | condições contratuais |
| `MonthlyCharges`, `TotalCharges` | valores cobrados |
| `Churn` | **variável-alvo** — Yes/No |

**Atenção:** a coluna `TotalCharges` está com tipo texto, não numérico — existe um motivo para isso. Investiguem antes de converter direto.
